```python
import pandas as pd
import numpy as np
import os
import glob
```

```python
def calculate_inner_brow_raiser_feature(data):
    inner_brow_raiser_feature = []
    baseline_Y21 = data['Y21'].mean()
    baseline_Y22 = data['Y22'].mean()
    baseline_Y23 = data['Y23'].mean()
    baseline_Y24 = data['Y24'].mean()
    baseline = (baseline_Y21 + baseline_Y22 + baseline_Y23 + baseline_Y24) / 4
    for index, row in data.iterrows():
        Y21 = row['Y21']
        Y22 = row['Y22']
        Y23 = row['Y23']
        Y24 = row['Y24']
        inner_brow_y = (Y21 + Y22 + Y23 + Y24) / 4
        inner_brow_raiser_feature.append(inner_brow_y - baseline)
    return inner_brow_raiser_feature

def calculate_outer_brow_raiser_feature(data):
    outer_brow_raiser_feature = []
    baseline_Y25 = data['Y25'].mean()
    baseline_Y26 = data['Y26'].mean()
    baseline_Y19 = data['Y19'].mean()
    baseline_Y20 = data['Y20'].mean()
    baseline = (baseline_Y25 + baseline_Y26 + baseline_Y19 + baseline_Y20) / 4
    for index, row in data.iterrows():
        Y21 = row['Y25']
        Y22 = row['Y26']
        Y23 = row['Y19']
        Y24 = row['Y20']
        outer_brow_raiser_y = (Y21 + Y22 + Y23 + Y24) / 4
        outer_brow_raiser_feature.append(outer_brow_raiser_y - baseline)
    return outer_brow_raiser_feature

def calculate_brow_lowerer(data):
    left_eyebrow_high = (data['X20'], data['Y20'])
    left_eyebrow_low = (data['X22'], data['Y22'])
    right_eyebrow_high = (data['X25'], data['Y25'])
    right_eyebrow_low = (data['X23'], data['Y23'])
    left_eyebrow_distance = np.sqrt((left_eyebrow_high[0] - left_eyebrow_low[0])**2 + (left_eyebrow_high[1] - left_eyebrow_low[1])**2)
    right_eyebrow_distance = np.sqrt((right_eyebrow_high[0] - right_eyebrow_low[0])**2 + (right_eyebrow_high[1] - right_eyebrow_low[1])**2)
    brow_lowerer_feature = (left_eyebrow_distance + right_eyebrow_distance) / 2
    return brow_lowerer_feature

def calculate_upper_lid_raiser(data):
    left_upper_lid_distance = ((np.sqrt((data['X38'] - data['X42'])**2 + (data['Y38'] - data['Y42'])**2)) + (np.sqrt((data['X39'] - data['X41'])**2 + (data['Y39'] - data['Y41'])**2))) / 2
    right_upper_lid_distance = ((np.sqrt((data['X44'] - data['X48'])**2 + (data['Y44'] - data['Y48'])**2)) + (np.sqrt((data['X45'] - data['X47'])**2 + (data['Y45'] - data['Y47'])**2))) / 2
    upper_lid_raiser = (left_upper_lid_distance + right_upper_lid_distance) / 2
    return upper_lid_raiser

def calculate_cheek_raiser(data):
    left_cheek_distance = np.sqrt((data['X49'] - data['X37'])**2 + (data['Y49'] - data['Y37'])**2)
    right_cheek_distance = np.sqrt((data['X55'] - data['X46'])**2 + (data['Y55'] - data['Y46'])**2)
    cheek_raiser_feature = (left_cheek_distance + right_cheek_distance) / 2

    return cheek_raiser_feature

def calculate_lid_tightener(data):
    left_eye_upper = (data['X38'], data['Y38'])
    left_eye_lower = (data['X42'], data['Y42'])
    right_eye_upper = (data['X44'], data['Y44'])
    right_eye_lower = (data['X48'], data['Y48'])
    left_eye_distance = np.sqrt((left_eye_upper[0] - left_eye_lower[0])**2 + (left_eye_upper[1] - left_eye_lower[1])**2)
    right_eye_distance = np.sqrt((right_eye_upper[0] - right_eye_lower[0])**2 + (right_eye_upper[1] - right_eye_lower[1])**2)
    lid_tightener_feature = (left_eye_distance + right_eye_distance) / 2
    return lid_tightener_feature

def calculate_nose_wrinkler(data):

    nose_bridge = (data['X28'], data['Y28'])
    nose_tip = (data['X34'], data['Y34'])
    nose_distance = np.sqrt((nose_bridge[0] - nose_tip[0])**2 + (nose_bridge[1] - nose_tip[1])**2)
    return nose_distance

def calculate_upper_lip_raiser(data):
    upper_lip = (data['X52'], data['Y52'])
    nose_tip = (data['X34'], data['Y34'])
    upper_lip_distance = np.sqrt((upper_lip[0] - nose_tip[0])**2 + (upper_lip[1] - nose_tip[1])**2)
    return upper_lip_distance

def calculate_nasolabial_furrow_deepener(data):
    left_nasolabial_furrow = (data['X33'], data['Y33'])
    right_nasolabial_furrow = (data['X35'], data['Y35'])
    nasolabial_furrow_distance = np.sqrt((left_nasolabial_furrow[0] - right_nasolabial_furrow[0])**2 + (left_nasolabial_furrow[1] - right_nasolabial_furrow[1])**2)
    return nasolabial_furrow_distance

def au12_lip_corner_puller(data):
    left_lip_corner = (data['X49'], data['Y49'])
    right_lip_corner = (data['X55'], data['Y55'])
    lip_corner_distance = np.sqrt((left_lip_corner[0] - right_lip_corner[0])**2 + (left_lip_corner[1] - right_lip_corner[1])**2)
    return lip_corner_distance


def au13_cheek_puffer(data):
    left_cheek = (data['X13'], data['Y13'])
    right_cheek = (data['X5'], data['Y5'])
    cheek_distance = np.sqrt((left_cheek[0] - right_cheek[0])**2 + (left_cheek[1] - right_cheek[1])**2)
    return cheek_distance

def au14_dimpler(data):
    left_dimple = (data['X49'], data['Y49'])
    right_dimple = (data['X55'], data['Y55'])
    dimple_distance = np.sqrt((left_dimple[0] - right_dimple[0])**2 + (left_dimple[1] - right_dimple[1])**2)
    return dimple_distance

def au15_lip_corner_depressor(data):
    lip_corner_distance = np.sqrt((data['Y56'] - data['Y58'])**2 + (data['Y60'] - data['Y58'])**2) / 2
    return lip_corner_distance

def au16_lower_lip_depressor(data):
    baseline1 = (data['Y58'].mean() - data['Y9'].mean())
    baseline2 = (data['Y67'].mean() - data['Y58'].mean())
    change_y = baseline2 / baseline1
    baseline_x = (data['X57'].mean() - data['X59'].mean())
    change_x = (data['Y57'] - data['Y59'] - baseline_x)
    lower_lip_distance = (change_x + change_y) / 2
    return lower_lip_distance

def au17_chin_raiser(data):
    chin = (data['X9'], data['Y9'])
    chin_distance = np.sqrt((chin[0] - data['X31'])**2 + (chin[1] - data['Y31'])**2)
    return chin_distance

def au18_lip_puckerer(data):
    upper_lip = (data['X49'].mean(), data['Y49'].mean())
    lower_lip = (data['X55'].mean(), data['Y55'].mean())
    lip_distance = np.sqrt((upper_lip[0] - lower_lip[0])**2 + (upper_lip[1] - lower_lip[1])**2)
    score = lip_distance - np.sqrt((data['X49'] - data['X55'])**2 + (data['Y49'] - data['Y55'])**2)
    return lip_distance

def au20_lip_stretcher(data):
    left_lip_corner = (data['X49'], data['Y49'])
    right_lip_corner = (data['X55'], data['Y55'])
    lip_corner_distance = np.sqrt((left_lip_corner[0] - right_lip_corner[0])**2 + (left_lip_corner[1] - right_lip_corner[1])**2)
    return lip_corner_distance

def au22_lip_funneler(data):
    upper_lip = (data['X52'], data['Y52'])
    lower_lip = (data['X58'], data['Y58'])
    lip_funneler_feature = np.sqrt((upper_lip[0] - lower_lip[0])**2 + (upper_lip[1] - lower_lip[1])**2)
    return lip_funneler_feature

def au23_lip_tightener(data):
    baseline1 = (data['X52'].mean(), data['Y52'].mean())
    baseline2 = (data['X63'].mean(), data['Y63'].mean())
    baseline3 = (data['X67'].mean(), data['X67'].mean())
    baseline4 = (data['X58'].mean(), data['X58'].mean())
    upper_lip = np.sqrt((baseline1[0] - baseline2[0])**2 + (baseline1[1] - baseline2[1])**2)
    lower_lip = np.sqrt((baseline3[0] - baseline4[0])**2 + (baseline3[1] - baseline4[1])**2)
    lip_tightener_feature = np.sqrt(upper_lip**2 / ((data['X52'] - data['X63'])**2 + (data['Y52'] - data['Y63'])**2)) + np.sqrt(lower_lip**2 / ((data['X58'] - data['X67'])**2 + (data['Y58'] - data['Y67'])**2))
    return lip_tightener_feature

def au24_lip_pressor(data):
    upper_lip = (data['X52'], data['Y52'])
    lower_lip = (data['X58'], data['Y58'])
    lip_pressor_feature = np.sqrt((upper_lip[0] - lower_lip[0])**2 + (upper_lip[1] - lower_lip[1])**2)
    return lip_pressor_feature

def au25_lips_part(data):
    upper_lip = (data['X52'], data['Y52'])
    lower_lip = (data['X58'], data['Y58'])
    lips_part_feature = np.sqrt((upper_lip[0] - lower_lip[0])**2 + (upper_lip[1] - lower_lip[1])**2)
    return lips_part_feature

def au26_jaw_drop(data):
    jaw_line = (data['X58'], data['Y58'])
    chin = (data['X9'], data['Y9'])
    jaw_drop_feature = np.sqrt((jaw_line[0] - chin[0])**2 + (jaw_line[1] - chin[1])**2)
    return jaw_drop_feature

def au27_mouth_stretch(data):
    left_lip_corner = (data['X52'], data['Y52'])
    right_lip_corner = (data['X58'], data['Y58'])
    mouth_stretch_feature = np.sqrt((left_lip_corner[0] - right_lip_corner[0])**2 + (left_lip_corner[1] - right_lip_corner[1])**2)
    return mouth_stretch_feature

def au28_lip_suck(data):
    upper_lip = (data['X52'], data['Y52'])
    lower_lip = (data['X58'], data['Y58'])
    lip_suck_feature = np.sqrt((upper_lip[0] - lower_lip[0])**2 + (upper_lip[1] - lower_lip[1])**2)
    return lip_suck_feature


def euclidean_distance(point1, point2):
    return np.sqrt((point1[0] - point2[0])**2 + (point1[1] - point2[1])**2)


def au41_lid_droop(data):
    upper_eyelid = (data['X39'], data['Y39'])  
    lower_eyelid = (data['X41'], data['Y41'])  
    pupil_center = ((data['X37'] + data['X40']) / 2, (data['Y37'] + data['Y40']) / 2)
    droop_distance = euclidean_distance(upper_eyelid, pupil_center)
    return droop_distance

def au42_slit(data):
    upper_eyelid = (data['X38'], data['Y38'])
    lower_eyelid = (data['X42'], data['Y42'])
    slit_distance = euclidean_distance(upper_eyelid, lower_eyelid)
    return slit_distance

def au43_eyes_closed(data):
    left_eye_upper = (data['X39'], data['Y39'])
    left_eye_lower = (data['X41'], data['Y41'])
    right_eye_upper = (data['X45'], data['Y45'])
    right_eye_lower = (data['X47'], data['Y47'])
    left_eye_distance = np.sqrt((left_eye_upper[0] - left_eye_lower[0])**2 + (left_eye_upper[1] - left_eye_lower[1])**2)
    right_eye_distance = np.sqrt((right_eye_upper[0] - right_eye_lower[0])**2 + (right_eye_upper[1] - right_eye_lower[1])**2)
    eyes_closed_feature = (left_eye_distance + right_eye_distance) / 2
    return eyes_closed_feature

def au44_squint(data):
    left_eye_outer = (data['X37'], data['Y37'])
    left_eye_inner = (data['X40'], data['Y40'])
    right_eye_outer = (data['X46'], data['Y46'])
    right_eye_inner = (data['X43'], data['Y43'])
    left_eye_distance = np.sqrt((left_eye_outer[0] - left_eye_inner[0])**2 + (left_eye_outer[1] - left_eye_inner[1])**2)
    right_eye_distance = np.sqrt((right_eye_outer[0] - right_eye_inner[0])**2 + (right_eye_outer[1] - right_eye_inner[1])**2)
    squint_feature = (left_eye_distance + right_eye_distance) / 2
    return squint_feature

def au45_blink(data):
    left_eye_upper = (data['X39'], data['Y39'])
    left_eye_lower = (data['X41'], data['Y41'])
    right_eye_upper = (data['X45'], data['Y45'])
    right_eye_lower = (data['X47'], data['Y47'])
    left_eye_distance = np.sqrt((left_eye_upper[0] - left_eye_lower[0])**2 + (left_eye_upper[1] - left_eye_lower[1])**2)
    right_eye_distance = np.sqrt((right_eye_upper[0] - right_eye_lower[0])**2 + (right_eye_upper[1] - right_eye_lower[1])**2)
    blink_feature = (left_eye_distance + right_eye_distance) / 2
    return blink_feature

def au46_wink(data):
    left_eye_upper = (data['X39'], data['Y39'])
    left_eye_lower = (data['X41'], data['Y41'])
    right_eye_upper = (data['X45'], data['Y45'])
    right_eye_lower = (data['X47'], data['Y47'])
    left_eye_distance = np.sqrt((left_eye_upper[0] - left_eye_lower[0])**2 + (left_eye_upper[1] - left_eye_lower[1])**2)
    right_eye_distance = np.sqrt((right_eye_upper[0] - right_eye_lower[0])**2 + (right_eye_upper[1] - right_eye_lower[1])**2)
    wink_feature = (left_eye_distance + right_eye_distance) / 2
    return wink_feature

```

```python

def compute_all_features(data):
    facial_features = {}
    facial_features['inner_brow_raiser'] = calculate_inner_brow_raiser_feature(data)
    facial_features['outer_brow_raiser'] = calculate_outer_brow_raiser_feature(data)
    facial_features['brow_lowerer'] = calculate_brow_lowerer(data)
    facial_features['upper_lid_raiser'] = calculate_upper_lid_raiser(data)
    facial_features['cheek_raiser_feature'] = calculate_cheek_raiser(data)
    facial_features['lid_tightener'] = calculate_lid_tightener(data)
    facial_features['nose_wrinkler'] = calculate_nose_wrinkler(data)
    facial_features['upper_lip_raiser'] = calculate_upper_lip_raiser(data)
    facial_features['nasolabial_furrow_deepener'] = calculate_nasolabial_furrow_deepener(data)
    facial_features['lip_corner_puller'] = au12_lip_corner_puller(data)
    facial_features['cheek_puffer'] = au13_cheek_puffer(data)
    facial_features['dimpler'] = au14_dimpler(data)
    facial_features['lip_corner_depressor'] = au15_lip_corner_depressor(data)
    facial_features['lower_lip_depressor'] = au16_lower_lip_depressor(data)
    facial_features['chin_raiser'] = au17_chin_raiser(data)
    facial_features['lip_puckerer'] = au18_lip_puckerer(data)
    facial_features['lip_stretcher'] = au20_lip_stretcher(data)
    facial_features['lip_funneler'] = au22_lip_funneler(data)
    facial_features['lip_tightener'] = au23_lip_tightener(data)
    facial_features['lip_pressor'] = au24_lip_pressor(data)
    facial_features['lips_part'] = au25_lips_part(data)
    facial_features['jaw_drop'] = au26_jaw_drop(data)
    facial_features['mouth_stretch'] = au27_mouth_stretch(data)
    facial_features['lip_suck'] = au28_lip_suck(data)
    facial_features['lid_droop'] = au41_lid_droop(data)
    facial_features['slit'] = au42_slit(data)
    facial_features['eye_closed'] = au43_eyes_closed(data)
    facial_features['squint'] = au44_squint(data)
    facial_features['blink'] = au45_blink(data)
    facial_features['wink'] = au46_wink(data)
    
    return facial_features
```

```python
def process_subjects(input_dir, output_dir):
    # List all subject directories in the input directory
    subject_folders = [folder for folder in os.listdir(input_dir) if os.path.isdir(os.path.join(input_dir, folder))]

    for subject in subject_folders:
        # Path to the subject's folder
        subject_folder_path = os.path.join(input_dir, subject)
        
        # Find all session subfolders
        subfolders = [subfolder for subfolder in os.listdir(subject_folder_path) if os.path.isdir(os.path.join(subject_folder_path, subfolder))]

        for subfolder in subfolders:
            # Path to the session subfolder
            subfolder_path = os.path.join(subject_folder_path, subfolder)
            
            # Find any file with "FL" in its name in this subfolder
            csv_files = glob.glob(os.path.join(subfolder_path, "*FL*.csv"))
            
            if csv_files:
                for csv_file in csv_files:
                    # Load the data
                    data = pd.read_csv(csv_file)

                    # Extract the frame numbers (first column)
                    frame_numbers = data.iloc[:, 0]

                    # Compute features using the function (assuming data excludes frame numbers)
                    features = compute_all_features(data.iloc[:, 1:])  # Exclude the first column (frame numbers) for feature computation

                    # Convert features to a DataFrame, including the frame numbers
                    features_df = pd.DataFrame(features)

                    # Add the frame numbers as the first column of the new DataFrame
                    features_df.insert(0, 'Frame', frame_numbers)

                    # Ensure the user/session output directory exists
                    user_output_folder = os.path.join(output_dir, subject, subfolder)
                    os.makedirs(user_output_folder, exist_ok=True)

                    # Save the file with the correct naming format: "User_FAU_Session.csv"
                    output_csv_path = os.path.join(user_output_folder, f'{subject}_FAU_{subfolder}.csv')
                    features_df.to_csv(output_csv_path, index=False)

                    print(f"✅ Processed {subject} ({subfolder}), output saved to {output_csv_path}")
            else:
                print(f"⚠️ No CSV file containing 'FL' found in subfolder {subfolder} of subject {subject}")

# Define input and output directories
input_dir = "Extracted_Features"
output_dir = "Extracted_Features"

# Run processing
process_subjects(input_dir, output_dir)
```
